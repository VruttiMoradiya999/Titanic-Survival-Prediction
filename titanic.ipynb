```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('titanic')
```

```python
df.columns
```

```python
df
```

```python
df.drop(['deck','embark_town',
         'alive','alone','class','who'],axis=1,inplace=True)
```

```python
df
```

```python
df['age'] = df['age'].fillna(df['age'].mean())
```

```python
df.info()
```

```python
df.dropna(subset=["embarked"],inplace=True)
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
```

```python
df.head()
```

```python
df['sex'] = le.fit_transform(df["sex"])
```

```python
df = pd.get_dummies(df, columns=['embarked'],drop_first=True, prefix='embarked', dtype=int)
```

```python
df.astype(int)
```

```python
X = df.drop(columns=["survived"])
y = df["survived"]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()

model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, classification_report , confusion_matrix
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test, y_pred)
```

```python
print(classification_report(y_test, y_pred))
```

```python
from sklearn. preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.fit_transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train_scaled,y_train)
```

```python
y_pred_knn = knn_model.predict(X_test)
```

```python
y_pred_knn
```

```python
accuracy_score(y_test,y_pred_knn)
```

```python
from sklearn.naive_bayes import GaussianNB

model_NB = GaussianNB()

model_NB.fit(X_train, y_train)
```

```python
y_pred_nb = model_NB.predict(X_test)
```

```python
accuracy_score(y_test,y_pred_nb)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_DT = DecisionTreeClassifier(random_state=42)
```

```python
model_DT.fit(X_train_scaled, y_train)
```

```python
y_pred_dtc = model_DT.predict(X_test_scaled)
```

```python
accuracy_score(y_test, y_pred_dtc)
```

```python
from sklearn.svm import SVC
model_svc = SVC(kernel = 'rbf')
model_svc.fit(X_train_scaled, y_train)
```

```python
y_pred_svc = model_svc.predict(X_test_scaled)
accuracy_score(y_test, y_pred_svc)
```

```python
import joblib
joblib.dump(model_svc, 'SVC_titanic.pkl')
joblib.dump(scaler, 'scaler.pkl')
joblib.dump(X.columns.to_list(), 'columns.pkl')
```
